## 3. Quick look at the data (2-3 real numbers)

The dataset contains 30,000 pages, of which 16,262 (54.2%) are labeled as declining. Reviewing all of these pages manually would be difficult, so prioritization is needed. Search volume has almost no linear correlation with actual impressions (0.001), and declining and growing pages have very similar median word counts (2,909 vs. 2,847.5). This suggests that simple rules based on one feature are not enough, and multiple signals should be considered when deciding which pages to review first.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 1. How many pages are declining?
declining_count = (df["trend_direction"] == "down").sum()
declining_pct = declining_count / len(df) * 100

# 2. Correlation between search volume and impressions
search_impression_corr = df["search_volume"].corr(df["impressions_90d"])

# 3. Median word count for declining vs growing pages
median_word_count = df.groupby("trend_direction")["word_count"].median()

print(f"Total pages: {len(df):,}")
print(f"Declining pages: {declining_count:,} ({declining_pct:.1f}%)")
print(f"Correlation between search volume and impressions: {search_impression_corr:.3f}")
print()
print("Median word count by trend:")
print(median_word_count)
```

## 4. Careful words: what I can and can't claim

I can claim that the available data can be used to identify and rank pages that show signals associated with decline or content opportunity. This ranking can support a human reviewer in deciding which pages to inspect first.
